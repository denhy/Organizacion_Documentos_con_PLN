```python
import os
from pdf2image import convert_from_path
import pytesseract
from pdfminer.high_level import extract_text

# 🔹 Ruta al ejecutable de Tesseract (ajústala si es diferente)
pytesseract.pytesseract.tesseract_cmd = r"C:\Program Files\Tesseract-OCR\tesseract.exe"


def has_selectable_text(pdf_path: str) -> bool:
    """Verifica si un PDF contiene texto seleccionable (no escaneado)."""
    try:
        text = extract_text(pdf_path)
        # Limpia el texto y verifica si hay contenido real
        clean_text = text.strip()
        return len(clean_text) > 50  # ajusta el umbral según tus documentos
    except Exception as e:
        print(f"⚠️ Error analizando PDF: {e}")
        return False


def extract_text_from_pdf(pdf_path: str) -> str:
    """Extrae texto del PDF: intenta texto directo y, si no hay, usa OCR."""
    if not os.path.exists(pdf_path):
        raise FileNotFoundError(f"No se encontró el archivo: {pdf_path}")

    print(f"📄 Analizando: {pdf_path}")

    # Paso 1: intentar extraer texto directamente
    if has_selectable_text(pdf_path):
        print("✅ El PDF tiene texto seleccionable. No se necesita OCR.")
        return extract_text(pdf_path).strip()

    # Paso 2: usar OCR si no hay texto seleccionable
    print("❌ No se encontró texto seleccionable. Aplicando OCR...")
    images = convert_from_path(pdf_path, dpi=300, poppler_path=r"C:\poppler\Library\bin")

    full_text = ""
    for i, img in enumerate(images):
        text = pytesseract.image_to_string(img, lang="spa")
        full_text += f"\n\n--- Página {i + 1} ---\n{text}"

    print("✅ OCR completado.")
    return full_text.strip()


if __name__ == "__main__":
    # 📎 Ejemplo de uso
    ruta_pdf = r"D:\Documentos\TRABAJOS_YO\UAEH\PRACTICAS PROFESIONALES\04-PPCC-Numero de cuenta.pdf"

    texto = extract_text_from_pdf(ruta_pdf)
    print("\n=== TEXTO EXTRAÍDO ===\n")
    print(texto[:2000])
```
